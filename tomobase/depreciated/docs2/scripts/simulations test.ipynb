```python
import numpy as np
from functools import partial
import pathlib
import os
from tomobase import phantoms, procedures


path = r"\\ematbyname\emat\Adrien\DIPSTER\TwinMode"
times = np.arange(1,101)
ko_1, e_1 = 0.4, 0.04
ko_2, e_2 = 0.7, 0.07

volume = phantoms.get_nanocage()
volume_deformed = volume
for t in times:
    if t < 50:
        ko, e = ko_1, e_1
    else:
        ko, e = ko_2, e_2
    volume_deformed = procedures.beamdamage(volume_deformed, knock_on=ko, elastic_deform=e)
    volume_deformed.write(pathlib.Path(os.path.join(path, f'beam_damage_dual_mode_{t:03d}.rec')))
```

```python
import numpy as np
from functools import partial
import pathlib
import os
from tomobase import phantoms, procedures
from tomobase.core.data_classes.images import Volume
from skimage.filters import threshold_otsu




path2 = r"\\ematbyname\emat\TimC\MeltTest"
path = r"\\ematbyname\emat\Adrien\DIPSTER\ResultsDIPSTER\NS2_220_byIndex_rot90_red-koopa-2-alltime"
times = np.arange(1,101)


volume = Volume.read(pathlib.Path(os.path.join(path, f'0data.rec')))
threshold = threshold_otsu(volume.data)
volume.xr.data = volume.data > threshold
volume_deformed = volume

for t in times:
    volume_deformed = procedures.melting(volume_deformed,  migration_probability=0.2, max_centroids=1)
    volume_deformed.write(pathlib.Path(os.path.join(path2, f'melting_{t:03d}.rec')))
    print(np.sum(volume_deformed.xr.data), np.count_nonzero(volume_deformed.xr.data))
```
