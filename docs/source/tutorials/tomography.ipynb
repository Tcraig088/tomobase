# Basic Tomography

This is a tutorial for performing basic tomography procedures using the tomobase library. Use this function in the command prompt to upload the documentation sphinx-autobuild docs/source docs/build --fresh-env

## Initialization

The following script initializes the libraries global settings allowing it to be used. This imports tomobase, sets the device management feautures and initializes the library.


```python
import logging
import tomobase
from tomobase import core # namespace for every component that does not require bootstrapping

core.logger.setLevel(logging.VERBOSE) # set logging level to verbose for detailed output, can be set to logging.INFO or logging.WARNING for less output
core.proxy.set_context(core.GPUContext.CUPY) #sets functions to use CUPY on GPU zero by default can be set to NUMPY

# bootstrap is necessary to load domain information 
# both Jupyter and qt environments are supported but disabled by default - needed if you want to use the interactive visualization features of tomobase.
# tomobase.bootstrap() will load the domain without visualization, useful for clean environments and import times.
tomobase.bootstrap(jupyter_enabled=True) 

from tomobase import procedures, images, tiltschemes
```

## Importing Data

Two alternative scripts are provided here. In the first a phantom is generated as a volume and converted into a sinogram. In the second, a file is read to import a sinogram. To see available phantoms, tiltschemes and read writes available see the [APIs](..\apis\index.md).



```python
import numpy as np

vol = procedures.phantoms.get_nanocage() 
angles = np.array(tiltschemes.Incremental()[0:71])
sino = procedures.project.forward_project(vol, angles, use_3D=True)

#sino.interactive.info()

sino = procedures.image_processing.scaling.normalize(sino)
sino = procedures.image_processing.misalignments.translational_misalignment(sino)
sino = procedures.image_processing.misalignments.poisson_noise(sino, rescale=10)
```

```python
# core.data_classes.Sinogram also works
vol = images.Sinogram.read()
```

To view your data, sino.interactive gives you an interactive view of the data.

## Data Processing

tomobase.procedures comes with numerous functions for manipulating image data. see the [APIs](..\apis\index.md), for more details. To understand how to customize your library with new functions see the [Extending Tomobase](extensions.ipynb) tutorial.

```python

# example use of cross correlation to align the sinogram
sino = procedures.align.shift_corrections.align_sinogram_xcorr(sino)
```

## Reconstruction

```python
vol = procedures.reconstruct.reconstruct_tvm(sino, tv_weight=0.005, weighted=False)
```
